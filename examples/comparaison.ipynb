```python
import scipy.stats as stats
import pandas as pd

```

```python
df = pd.read_csv("C:\\Users\\lucie\\G-n-rateur-d-intention\\Générateur d'intention\\data\\generated_instructions_with_intention_explanation_full_benevolence_gender_woman_gemma_QA_rated.csv")
df.head()
```

```python
df['LLMrating']=df['LLMrating'].apply(lambda x: 'High' if 'High' in x else ('Medium' if x == 'Medium' else 'Low'))
df.head()
```

```python
# Example data

# Calculate Spearman Correlation
for col in ['LLMrating', 'LLMrating']:
    rho, p_val = stats.spearmanr(df['benevolence'], df[col])
    print(f"{col} -> Rho: {rho:.2f}, p-value: {p_val:.4f}")
```

```python
import seaborn as sns
sns.histplot(data=df, x='LLMrating', hue='benevolence', multiple='stack', bins=10)
```
